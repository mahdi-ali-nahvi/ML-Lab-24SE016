## Task 7 · Climate of five cities (real data, PMD 1991–2020 normals)
Using `tmax` (mean daily maximum °C) and `rain` (mm), rows = `cities`, columns = `months`:
- **(a)** Shape of `tmax`.
- **(b)** Annual mean maximum temperature of each city.
- **(c)** Hottest month of each city (use `argmax` + `months`).
- **(d)** 5-city average for each month.
- **(e)** Number of months above 35 °C in each city.
- **(f)** Lahore's temperatures in °F (`F = C × 9/5 + 32`).
- **(g)** Monsoon (Jul–Sep) rain of each city by slicing columns.
- **(h)** The wettest city in the monsoon, and what % of each city's yearly rain falls in the monsoon.

**Interpret:** in 1–2 lines, which result surprised you, and why?

```python
# T7: your code here
print("T7a: shape of tmax =", tmax.shape)                                   # (a)
print("T7b: annual mean max temp:", dict(zip(cities.tolist(), np.round(tmax.mean(axis=1), 2).tolist())))  # (b)
print("T7c: hottest month:", dict(zip(cities.tolist(), months[tmax.argmax(axis=1)].tolist())))             # (c)
print("T7d: 5-city monthly average:", dict(zip(months.tolist(), np.round(tmax.mean(axis=0), 2).tolist())))  # (d)
print("T7e: months above 35 C:", dict(zip(cities.tolist(), (tmax > 35).sum(axis=1).tolist())))  # (e)
lahore_f = tmax[0] * 9/5 + 32                                                # (f)
print("T7f: Lahore in F:", np.round(lahore_f, 1))
monsoon = rain[:, 6:9].sum(axis=1)                                           # (g) Jul-Sep
print("T7g: monsoon rain (mm):", dict(zip(cities.tolist(), np.round(monsoon, 1).tolist())))
print("T7h: wettest city in monsoon =", cities[monsoon.argmax()], "-", round(monsoon.max(), 1), "mm")  # (h)
pct = monsoon / rain.sum(axis=1) * 100
print("T7h: % of yearly rain in monsoon:", dict(zip(cities.tolist(), np.round(pct, 1).tolist())))
```

**Interpretation (T7):** I was surprised that Karachi has the highest annual mean maximum (32.5 °C), yet it never peaks like the inland cities. Its hottest month is only 35.7 °C, because the sea keeps it moderate all year. Islamabad gets far more monsoon rain (763 mm) than Lahore (454 mm). Quetta is hottest in July, later than the others, and gets only 6 % of its rain in the monsoon. Note: Lahore (May/Jun = 38.9) and Karachi (May/Jun = 35.7) have ties, and `argmax` returns the first month.

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9: your code here
after = np.clip(marks + 5, 0, 100)                                   # (a)
print("T9a: before =", marks)
print("     after  =", after)
print("T9b: pass % after grace =", np.mean(after >= 50) * 100, "%  (before grace:", np.mean(marks >= 50) * 100, "%)")  # (b)
only_grace = np.where((marks < 50) & (after >= 50))[0]               # (c)
print("T9c: passed only because of grace -> indices", only_grace, "| marks before:", marks[only_grace])
z = (after - after.mean()) / after.std()                             # (d)
print("T9d: z-scores =", np.round(z, 2))
```
