## Task 6
**What is the main difference between a tuple and a list?**

**Answer (in my own words):** The main difference is **mutability**. A list (`[ ]`) is mutable, so after creating it I can change items, add new ones or remove them. A tuple (`( )`) is immutable, so once it is created its contents can never be changed, and trying to do so raises a `TypeError`. Because of this, tuples are good for fixed values such as `(rows, cols)`, while lists are used when the data needs to change.

The cell below demonstrates it.

```python
my_list = [1, 2, 3]
my_list[0] = 10          # works: lists are mutable
my_list.append(4)
print("Q6: list after changes ->", my_list)

my_tuple = (1, 2, 3)
try:
    my_tuple[0] = 10     # fails: tuples are immutable
except TypeError as e:
    print("Q6: tuple error ->", e)
```

## Task 9
**Function: count how many times "dog" occurs in a string (case-insensitive).**

```python
def count_dog(text):
    """Count occurrences of 'dog' in text, ignoring case."""
    return text.lower().count("dog")

print("Q9:", count_dog("This dog runs faster than the other dog"))
print("Q9:", count_dog("The Dog and the dog"))
print("Q9:", count_dog("No animals here"))
```

## Task 10
**`caught_speeding(speed, is_birthday)`**
- Speed ≤ 60 → `"No Ticket"`
- Speed 61–80 → `"Small Ticket"`
- Speed ≥ 81 → `"Big Ticket"`
- On your birthday, speeds are allowed to be 5 higher in all cases (so the limits shift by 5).

```python
def caught_speeding(speed, is_birthday):
    """Return the ticket type for a given speed.
    On a birthday, 5 extra mph is allowed, so we subtract 5 from the speed first."""
    if is_birthday:
        speed -= 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print("Q10:", caught_speeding(81, True))    # 76 -> Small Ticket
print("Q10:", caught_speeding(81, False))   # Big Ticket
print("Q10:", caught_speeding(60, False))   # No Ticket
print("Q10:", caught_speeding(65, True))    # 60 -> No Ticket (edge case)
print("Q10:", caught_speeding(86, True))    # 81 -> Big Ticket (edge case)
```

## Bonus 1
**Make `count_dog` also count `"dog,"` and `"dog!"` (punctuation).**

Each word is stripped of surrounding punctuation before comparing, so `dog,` `dog!` and `Dog.` all count.

```python
import string

def count_dog_punct(text):
    """Count the word 'dog' (case-insensitive), even with punctuation attached."""
    count = 0
    for word in text.lower().split():
        if word.strip(string.punctuation) == "dog":
            count += 1
    return count

print("Bonus 1:", count_dog_punct("Dog, dog! This dog? A dog. Dogs bark."))   # 4 (Dogs is not 'dog')
print("Bonus 1:", count_dog_punct("The dog, the Dog! and my dog"))           # 3
```

## Bonus 2
**Write `is_palindrome(word)` using slicing: `word[::-1]`.**

```python
def is_palindrome(word):
    """Return True if word reads the same forwards and backwards (case-insensitive)."""
    word = word.lower()
    return word == word[::-1]

print("Bonus 2:", is_palindrome("madam"))
print("Bonus 2:", is_palindrome("Level"))
print("Bonus 2:", is_palindrome("python"))
```
